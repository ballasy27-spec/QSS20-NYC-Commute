# 01 — Clean Data

This notebook cleans the data and creates the Bronx–Manhattan sample used for analysis.

```python
import pandas as pd
```

```python
def keep_boroughs(data):
    return data[data["Borough"].isin(["Bronx", "Manhattan"])]
```

```python
df = pd.read_csv("../data/acs2015_nyc_tracts_raw.csv")
```

```python
print(df.shape)
print(df["Borough"].value_counts())
```

```python
df = df.rename(columns={
    "DP03_0025E": "MeanCommute",
    "DP03_0021PE": "Transit",
    "DP03_0062E": "Income",
    "DP03_0128PE": "Poverty",
    "DP03_0009PE": "Unemployment",
    "DP03_0024PE": "WorkAtHome"
})
```

```python
df = df[[
    "Borough",
    "CensusTract",
    "MeanCommute",
    "Transit",
    "Income",
    "Poverty",
    "Unemployment",
    "WorkAtHome"
]]
```

```python
df = keep_boroughs(df)

df.shape
```

```python
columns = [
    "MeanCommute",
    "Transit",
    "Income",
    "Poverty",
    "Unemployment",
    "WorkAtHome"
]

df[columns] = df[columns].apply(pd.to_numeric, errors="coerce")
```

```python
df = df.dropna()

df = df[
    (df["MeanCommute"] >= 0) &
    (df["Transit"] >= 0) &
    (df["Income"] >= 0) &
    (df["Income"] != 250001) &
    (df["Poverty"] >= 0) &
    (df["Unemployment"] >= 0) &
    (df["WorkAtHome"] >= 0)
]
```

```python
print(df.shape)
print(df["Borough"].value_counts())
```

```python
df.to_csv("../data/acs2015_bronx_manhattan_clean.csv", index=False)
```
