# 00 — Pull Data

The notebook pulls 2015 American Community Survey census-tract data for New York City directly from the U.S. Census Bureau.

My API code for accessing the Census is: sent to email. Just copy and paste it


```python
import pandas as pd
```

```python
def get_borough(county):
    if county == "005":
        return "Bronx"
    elif county == "047":
        return "Brooklyn"
    elif county == "061":
        return "Manhattan"
    elif county == "081":
        return "Queens"
    elif county == "085":
        return "Staten Island"
```

## Data Source

The data come from the U.S. Census Bureau's 2015 American Community Survey 5-year estimates. I use census-tract measures of commute time, public-transit use, median household income, poverty, unemployment, and working from home.

```python
api_key = input("Enter your Census API key: ")
```

```python
url = (
    "https://api.census.gov/data/2015/acs/acs5/profile"
    "?get=NAME,DP03_0025E,DP03_0021PE,DP03_0062E,DP03_0128PE,DP03_0009PE,DP03_0024PE"
    "&for=tract:*"
    "&in=state:36"
    "&key=" + api_key
)
```

```python
df = pd.read_json(url, dtype=str)
```

```python
df.columns = df.iloc[0]
df = df.drop(0)
df = df.reset_index(drop=True)
```

```python
nyc_counties = ["005", "047", "061", "081", "085"]

df = df[df["county"].isin(nyc_counties)].copy()
```

```python
df["Borough"] = df["county"].apply(get_borough)
```

```python
df["CensusTract"] = df["state"] + df["county"] + df["tract"]
```

```python
print("Number of NYC census tracts:", len(df))
print()
print(df["Borough"].value_counts())
```
